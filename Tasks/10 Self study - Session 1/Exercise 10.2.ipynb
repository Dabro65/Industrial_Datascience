Exercise 10.2 (Working with JupyterLab)  

JupyterLab is a web-based interactive development environment (IDE) for Jupyter Notebooks. It provides a flexible interface for interactive data exploration and reporting.

Launch JupyterLab from your uv environment by adding it to your environment first or alternatively, using VS Code, you can also use the Jupyter extension instead.
Create a new notebook called ai4i_explore.ipynb.
Load the AI4I dataset into a pandas DataFrame using a code cell. Keep only the columns UDI (unique identifier which can be used as 
axis for plotting), Torque, Rotational speed and Air temperature.
Display the first 10 rows and print basic statistics.
Create a line plot of Torque over UDI using matplotlib.
Experiment with different cell types: code, markdown, and raw cells.
Export the notebook as HTML and PDF.
After this exercise you should be able to:

Create and organize notebooks in JupyterLab.
Differentiate between code, markdown, and raw cells.
Export notebooks to other formats (HTML, PDF).
Structure a notebook with self-contained, readable code cells.
