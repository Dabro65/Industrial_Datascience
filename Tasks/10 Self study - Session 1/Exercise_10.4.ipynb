Exercise 10.4 (Generating a report with Quarto)
Quarto is an open-source scientific and technical publishing system. It allows you to create dynamic documents,
reports, and presentations from your code and narrative text. The material of this course itself is rendered using
Quarto.
1. Create a Quarto Markdown ( .qmd ) file in your project.
2. Embed the code and plots from the previous exercise into the Quarto document.
3. Use code chunk options to control output (e.g., echo=FALSE to hide code, fig-width to set figure dimensions).
4. Render the document as HTML and PDF.
5. Explore collapsing code chunks in your document (see Quarto docs).
6. Quarto also allows to render Jupyter notebook files ( .ipynb ) directly to HTML and PDF (see Quarto docs):
Prepare the notebook you created in the previous exercise for rendering with Quarto by adding additional
information that makes a nice presentation.
Render the notebook you created as HTML and PDF using Quarto.
After this exercise you should be able to:
1. Create a Quarto document that integrates code, output, and narrative.
2. Render Quarto documents to HTML and PDF formats.
