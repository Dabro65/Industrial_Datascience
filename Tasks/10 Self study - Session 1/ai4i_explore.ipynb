```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# URL zum AI4I 2020 Predictive Maintenance Dataset.
# Der Datensatz enthält simulierte Sensordaten einer industriellen Maschine.
url = "https://archive.ics.uci.edu/static/public/601/ai4i+2020+predictive+maintenance+dataset.zip"

# Lädt die CSV-Datei aus dem ZIP-Archiv in einen Pandas DataFrame.
# Ein DataFrame ist eine tabellarische Datenstruktur mit Zeilen und Spalten.
df = pd.read_csv(url)

# Wählt die vier für diese Aufgabe benötigten Spalten aus.
df = df[["UDI", "Torque [Nm]", "Rotational speed [rpm]", "Air temperature [K]"]]

print("Erste 10 Zeilen:")

# Zeigt die ersten zehn Zeilen des DataFrames an.
print(df.head(10))

# Berechnet grundlegende statistische Kennwerte der numerischen Spalten.
# Dazu gehören Anzahl, Mittelwert, Standardabweichung,
# Minimum, Maximum und die Quartile (25 %, 50 %, 75 %).
print("\nStatistische Kennwerte:")
print(df.describe())
```

```python
import matplotlib.pyplot as plt     # Importiert Matplotlib

# Erstellt eine neue Abbildung mit einer Breite von 12 und Höhe von 5 Zoll.
plt.figure(figsize=(12, 5))

# Zeichnet das Drehmoment über der eindeutigen Datensatznummer UDI.
# Jeder Datenpunkt wird über eine Linie mit dem nächsten verbunden.
plt.plot(df["UDI"], df["Torque [Nm]"])

# Fügt einen Titel und die Achsenbeschriftungen hinzu.
plt.title("Torque over UDI")
plt.xlabel("UDI")
plt.ylabel("Torque [Nm]")

# Passt die Abstände automatisch an, damit Beschriftungen nicht abgeschnitten werden.
plt.tight_layout()

# Zeigt das fertige Diagramm an.
plt.show()
```
