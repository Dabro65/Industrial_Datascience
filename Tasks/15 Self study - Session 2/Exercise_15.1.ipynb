4. Load the dataset into your environment and inspect its structure: number of rows, columns, data types, and any obvious issues.

```python
import pandas as pd

# Laden des Datensatzes aus dem Projektorder
df = pd.read_csv("XAI_Drilling_Dataset.csv")

# Anzeige der ersten 5 Zeilen des Datensatzes
display(df.head())

print("Zeilen:", df.shape[0])
print("Spalten:", df.shape[1])

# Anzeige Datentypen und fehlende Werte
df.info()

# Doppelte Zeilen und IDs prüfen
print("Doppelte Zeilen:", df.duplicated().sum())
print("Doppelte IDs:", df["ID"].duplicated().sum())

# Wertebereiche der numerischen Spalten prüfen
display(df.describe())

# Ausprägungen der kategorialen Spalten prüfen
print("Material:", df["Material"].unique())
print("Drill Bit Type:", df["Drill Bit Type"].unique())
```

Der Datensatz enthält 20.000 Beobachtungen und 15 Spalten. Dazu gehören eine ID, sieben numerische Prozessgrößen, die beiden kategorialen Merkmale Material und Drill Bit Type sowie fünf binäre Fehlerindikatoren.

Die numerischen Spalten sind als int64 beziehungsweise float64 gespeichert. Material und Drill Bit Type liegen als Text vor. Alle Spalten enthalten jeweils 20.000 nicht fehlende Werte, sodass keine fehlenden Werte vorhanden sind.

5. Identify the target variable(s) and any class imbalances.

```python
# Zielspalten für die Analyse mit den Ausfällen
target_cols = ["Main Failure", "BEF", "CCF", "FWF", "WDF"]

for target in target_cols:
    print(target)

    # Anzahl und Anteil der Ausfälle in der Zielspalte berechnen
    counts = df[target].value_counts().sort_index()
    percentages = df[target].value_counts(normalize=True).sort_index() * 100

    # Ausgabe dr Daten
    display(pd.DataFrame({
        "Anzahl": counts,
        "Anteil [%]": percentages
    }))
```

Die Zielvariablen sind Main Failure sowie die einzelnen Fehlerindikatoren BEF, CCF, FWF und WDF. Alle sind binär codiert: 0 bedeutet kein entsprechender Fehler, 1 bedeutet Fehler.

Von insgesamt 20.000 Beobachtungen sind 1.001 als Main Failure gekennzeichnet, entsprechend 5,005 %. Die Untergruppen enthalten 215 BEF-Fälle (1,075 %), 344 CCF-Fälle (1,72 %), 278 FWF-Fälle (1,39 %) und 300 WDF-Fälle (1,5 %).

Damit sind die Klassen deutlich unausgeglichen. Ein Modell, das immer keinen allgemeinen Fehler vorhersagt, würde bereits 94,995 % Accuracy erreichen, aber keinen einzigen Ausfall erkennen. Zur Bewertung sind deshalb zusätzlich Precision, Recall und F1-Score sinnvoll.

6. Check the consistency of the labels: does the main failure flag coincide with the union of the subgroup failure flags? Are there samples with more than one subgroup failure? In case you find any, how would you handle such inconsistencies when training a model?

```python
# Untergruppenfehler-Spalten
subgroup_cols = ["BEF", "CCF", "FWF", "WDF"]

# Anzahl der Untergruppenfehler pro Beobachtung
failure_count = df[subgroup_cols].sum(axis=1)

# Erwarteter Hauptindikator: mindestens ein Untergruppenfehler
expected_main_failure = (failure_count > 0).astype(int)

# Widersprüche zwischen Hauptindikator und Untergruppen
inconsistent = df["Main Failure"] != expected_main_failure

print("Widersprüchliche Kennzeichnungen:", inconsistent.sum())

display(
    df.loc[inconsistent, ["ID", "Main Failure"] + subgroup_cols]
)

# Beobachtungen mit mehreren Untergruppenfehlern
multiple_failures = failure_count > 1

print("Beobachtungen mit mehreren Untergruppenfehlern:", multiple_failures.sum())

display(
    df.loc[multiple_failures, ["ID", "Main Failure"] + subgroup_cols]
)
```

Die Prüfung ergibt keine Widersprüche zwischen Main Failure und den Untergruppen BEF, CCF, FWF und WDF. Der Hauptindikator entspricht bei allen Beobachtungen der Vereinigung der Untergruppen: Sobald mindestens ein Untergruppenfehler vorliegt, ist Main Failure gleich 1.

Bei 114 Beobachtungen treten mehrere Untergruppenfehler gleichzeitig auf. Das entspricht 0,57 % des gesamten Datensatzes. Diese Mehrfachfehler sind keine inkonsistenten Kennzeichnungen und müssen nicht entfernt werden.

Für das Modelltraining können wir für jede Fehlerart ein eigenes binäres Modell verwenden. Dabei bleiben alle zutreffenden Fehlerindikatoren erhalten. Eine Einteilung in ausschließlich gegenseitig ausschließende Fehlerklassen wäre dagegen ungeeignet.

Da keine widersprüchlichen Kennzeichnungen gefunden wurden, ist aufgrund dieser Prüfung keine Korrektur der Labels erforderlich.
